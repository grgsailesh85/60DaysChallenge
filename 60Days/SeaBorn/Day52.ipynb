```python
import seaborn as sns
import matplotlib.pyplot as plt 
import pandas as pd
```

```python
sns.set_style("darkgrid")
sns.set_context("notebook", font_scale = 1.0)
```

```python
tips = sns.load_dataset("tips")
```

```python
# Plot Histogram and KDE
```

```python
plt.figure(figsize = (8,3))
sns.histplot(
    tips["tip"], 
)
plt.title("Histogram and KDE Plot of Total Bill")
plt.show()
```

```python
# Scatter Plot with HUe and Size
```

```python
plt.figure(figsize = (10, 5))
print(tips)
```

```python
sns.scatterplot(
    data = tips,
    x = "total_bill",
    y = "tip",
    hue = "day",
    size = "size",
    sizes = (20, 240)
)
plt.title("Total-Bill Vs Tip (hue = Day, Size = Order Size)")
plt.show()
```

```python
# Box + Swaarm Plot
```

```python
plt.figure(figsize = (10, 5))
sns.boxplot(
    data = tips,
    x = "day",
    y = "total_bill"
)
sns.swarmplot(
    data = tips,
    x = "day",
    y = "total_bill",
    color = "k",
    alpha = 0.4
)
plt.title("BoxPlot with swarm overlay")
plt.show()
```
