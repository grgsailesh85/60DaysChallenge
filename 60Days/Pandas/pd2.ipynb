```python
import pandas as pd
```

```python
data = {
    "A" : [10, 30, 50, 10],
    "B" : [5, 40, 60, 7],
    "C" : [10, 30, 20, 90],
    "D" : [20, 50, 10, 80]
}
```

```python
df = pd.DataFrame(data)
df
```

```python
# Basic Arithmetic Operations 
```

```python
# Addition, Subtraction, Multiplication, DIvision, Modulus
df["A + B"] = df["A"] + df["B"]
df["A - B"] = df["A"] - df["B"]
df["A * B"] = df["A"] * df["B"]
df["A / B"] = df["A"] / df["B"]
df["A mod B"] = df["A"] % df["B"]

```

```python
df
```

```python
# Statistics 
```

```python
df.mean()
```

```python
df.min()
```

```python
df.max()
```

```python
# Element wise Operations
```

```python
import numpy as np
df["Sq_C"] = np.sqrt(df["C"])
df["Log_A"] = np.log(df["A"])
df["Exp_B"] = np.exp(df["B"])
df["Sin_D"] = np.sin(df["D"])
df
```

```python
# apply()
def sq(x):
    return x ** 2

df["Square A"] = df["A"].apply(sq)
df
```

```python
# Column Wise and Row Wise Operations
```

```python
df["row_sum"] = df.sum(axis=1)
df["col_sum"] = df.sum(axis=0)
df
```

```python
col_sum = df.sum(axis=0)
col_sum_arr = np.array(col_sum, dtype="int32")
col_sum_arr
```
