```python
import pandas as pd
```

```python
df = pd.read_csv("sample_dirty_dataset.csv")
df
```

```python
# Suppose I have Dataset
# 1 -> In tht dataset I have some missing values
# 2 -> I have to fill them 
# 3 -> Before filling them we check for outliers
#    a -> If there are outliers we firstly remove them
# 4 -> We fill the missing values
# 5 -> fill them with mean (since there are no outliers now )   

```

```python
# Remove outliers
def remove_outliers_IQR(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]

```

```python
df_clean = (
    df
    .pipe(remove_outliers_IQR, column = "Age")
    .pipe(remove_outliers_IQR, column = "Salary")
)
df_clean
```

```python
# What about string values
# To fill up the missing data in case of strings
# 1 -> You can either keep them unknown -> used in analysis
# 2 -> You can replace them with the maximum repeated value(mode{mo}) -> used in ML
```

```python
# Fill the unknown values -> strings
df_clean["Department"] = df_clean["Department"].fillna("unknown")
df_clean
```

```python
# Remove duplicates
df_clean.drop_duplicates(inplace = True)
df_clean
```

```python
# Making Data consistent
df_clean["Gender"] = df["Gender"].replace({
    "Male" : "M",
    "Female" : "F"
})
df_clean
```

```python
df_clean.to_csv("Cleaned_data.csv", index = False)
df_clean
```
